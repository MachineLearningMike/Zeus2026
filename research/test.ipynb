```python
def best_segment_lengths_increasing(N, S):
    """
    Return a sequence of segment lengths (non-decreasing, sum=N)
    maximizing number of trailing equal segments.
    """
    best_seq = None
    best_trailing = 0
    
    def dfs(seq, remaining, remaining_count):
        nonlocal best_seq, best_trailing
        if remaining_count == 0:
            if remaining != 0:
                return
            # compute trailing uniform length
            last_val = seq[-1]
            count = 0
            for x in reversed(seq):
                if x == last_val:
                    count += 1
                else:
                    break
            if count > best_trailing:
                best_trailing = count
                best_seq = seq[:]
            return
        # next value ≥ previous (or 1 if seq empty)
        start = seq[-1] if seq else 1
        max_val = remaining - (remaining_count - 1)  # leave at least 1 for each remaining
        for val in range(start, max_val + 1):
            dfs(seq + [val], remaining - val, remaining_count - 1)
    
    dfs([], N, S)
    return best_seq

# Examples
print(best_segment_lengths_increasing(13, 4))  # [2, 3, 4, 4]
print(best_segment_lengths_increasing(13, 6))  # [1, 2, 2, 2, 3, 3]
```

```python
import math

def best_segment_lengths_decreasing(N, S):
    """
    Non-increasing segment lengths (earlier >= later),
    maximizing trailing uniformity, with more balanced distribution.
    """
    best_seq = None
    best_trailing = 0

    def dfs(seq, remaining, remaining_count):
        nonlocal best_seq, best_trailing
        if remaining_count == 0:
            if remaining != 0:
                return
            # trailing uniform length
            last_val = seq[-1]
            count = 0
            for x in reversed(seq):
                if x == last_val:
                    count += 1
                else:
                    break
            if count > best_trailing:
                best_trailing = count
                best_seq = seq[:]
            return
        # next value ≤ previous (or ceiling of even distribution if seq empty)
        start = seq[-1] if seq else math.ceil(remaining / remaining_count)
        min_val = 1
        for val in range(start, min_val-1, -1):
            if val * remaining_count < remaining:
                continue
            dfs(seq + [val], remaining - val, remaining_count - 1)

    dfs([], N, S)
    return best_seq

# Examples
print(best_segment_lengths_decreasing(13, 4))  # balanced, non-increasing
print(best_segment_lengths_decreasing(13, 6))  # balanced, non-increasing
```

```python
import math

def best_segments_increasing(N, S):
    """
    Return actual segments of numbers 0..N-1
    based on non-decreasing lengths maximizing trailing uniformity.
    Mirrors the logic of best_segments_decreasing.
    """
    # Step 1: compute segment lengths
    def best_segment_lengths_increasing(N, S):
        best_seq = None
        best_trailing = 0

        def dfs(seq, remaining, remaining_count):
            nonlocal best_seq, best_trailing
            if remaining_count == 0:
                if remaining != 0:
                    return
                # trailing uniform length
                last_val = seq[-1]
                count = 0
                for x in reversed(seq):
                    if x == last_val:
                        count += 1
                    else:
                        break
                if count > best_trailing:
                    best_trailing = count
                    best_seq = seq[:]
                return

            # Mirror logic: next value ≥ previous (or floor of even distribution if seq empty)
            start = seq[-1] if seq else max(1, remaining // remaining_count)
            max_val = remaining - (remaining_count - 1)  # leave at least 1 for remaining segments
            for val in range(start, max_val + 1):
                dfs(seq + [val], remaining - val, remaining_count - 1)

        dfs([], N, S)
        return best_seq

    lengths = best_segment_lengths_increasing(N, S)

    # Step 2: build actual segments
    segments = []
    start = 0
    for length in lengths:
        segments.append(list(range(start, start + length)))
        start += length
    return segments

# Examples
print(best_segments_increasing(13, 4))  # [2,3,4,4] lengths → segments of numbers
print(best_segments_increasing(13, 6))  # [1,2,2,2,3,3] lengths → segments of numbers
```

```python
import math

def best_segments_decreasing(N, S):
    """
    Return actual segments of numbers 0..N-1
    based on non-increasing lengths maximizing trailing uniformity.
    """
    # Step 1: compute segment lengths
    def best_segment_lengths_decreasing(N, S):
        best_seq = None
        best_trailing = 0

        def dfs(seq, remaining, remaining_count):
            nonlocal best_seq, best_trailing
            if remaining_count == 0:
                if remaining != 0:
                    return
                # trailing uniform length
                last_val = seq[-1]
                count = 0
                for x in reversed(seq):
                    if x == last_val:
                        count += 1
                    else:
                        break
                if count > best_trailing:
                    best_trailing = count
                    best_seq = seq[:]
                return
            # next value ≤ previous (or ceiling of even distribution if seq empty)
            start = seq[-1] if seq else math.ceil(remaining / remaining_count)
            min_val = 1
            for val in range(start, min_val-1, -1):
                if val * remaining_count < remaining:
                    continue
                dfs(seq + [val], remaining - val, remaining_count - 1)

        dfs([], N, S)
        return best_seq

    lengths = best_segment_lengths_decreasing(N, S)

    # Step 2: build actual segments
    segments = []
    start = 0
    for length in lengths:
        segments.append(list(range(start, start + length)))
        start += length
    return segments

# Examples
print(best_segments_decreasing(13, 4))
print(best_segments_decreasing(13, 6))
```

```python
def check_consistency(n, s):
    inc = best_segments_increasing(n, s)
    dec = best_segments_decreasing(n, s)
    def assert_full(A):
        sum = []
        for a in A:
            sum = sum + a
        assert len(A) == s
        assert len(sum) == n
        assert set(sum) == set(range(n))
    assert_full(inc)
    assert_full(dec)
    inc_l = [len(a) for a in inc]
    dec_l = [len(a) for a in dec]
    print(inc_l)
    print(dec_l)
    assert inc_l == dec_l[::-1]
    print(inc)
    print(dec)

for n in range(1, 30):
    for s in range(1, n):
        check_consistency(n, s)
```

```python
def grouped_indices(N, G):
    first = N % G or G
    return [list(range(0, first))] + [
        list(range(i, i + G)) for i in range(first, N, G)
    ]
```

```python
import torch

# toy example functions
def f(a):            # a -> b
    return 2 * a

def g(b):            # b -> c
    return (b ** 2).sum(dim=-1, keepdim=True)  # shape compatible with grad_out

# prepare inputs
a = torch.randn(3, 4, requires_grad=True)
b = f(a)             # b is non-leaf but part of graph
c = g(b)

# suppose we have an upstream gradient for c (same shape as c)
c_grad = torch.ones_like(c) * 0.5

# 1) compute gradient of c w.r.t. b (b_grad)
#    this computes d(c)/d(b) * c_grad and returns gradient for b only
b_grad, = torch.autograd.grad(outputs=c, inputs=b, grad_outputs=c_grad, retain_graph=True, create_graph=False)
# b_grad is a Tensor same shape as b

# 2) compute gradient of b w.r.t. a using b_grad as upstream
a_grad, = torch.autograd.grad(outputs=b, inputs=a, grad_outputs=b_grad, retain_graph=False, create_graph=False)
# a_grad is a Tensor same shape as a

print("b_grad.shape:", b_grad.shape)
print("a_grad.shape:", a_grad.shape)
```

```python

import torch
from torch.distributions import Categorical

logits = torch.tensor([[1.0, 2.0, 3.0],
                       [0.3, 0.3, 0.4]])
dist = Categorical(logits=logits)   # logits: (batchs, actions)
dist.sample()     # Samples one per row (batch), 
# tensor([2, 2])
dist.probs  # (batchs, actions)
# tensor([[0.0900, 0.2447, 0.6652],
#         [0.3220, 0.3220, 0.3559]])
torch.log(dist.probs)
# tensor([[-2.4076, -1.4076, -0.4076],
#         [-1.1331, -1.1331, -1.0331]])
dist.logits # (batchs, actions)
# tensor([[-2.4076, -1.4076, -0.4076],
#         [-1.1331, -1.1331, -1.0331]])
torch.softmax(dist.logits, dim=-1)
# tensor([[0.0900, 0.2447, 0.6652],
#         [0.3220, 0.3220, 0.3559]])
dist.log_prob(torch.tensor(2))  # action 2 in each batch
# tensor([-0.4076, -1.0331])
dist.probs[1]
# tensor([0.3220, 0.3220, 0.3559])
dist.log_prob(torch.tensor(1))  # action 1 in each batch
# tensor([-1.4076, -1.1331])
dist.log_prob(torch.tensor(1))[1]   # log_prob(action idx)
# tensor(-1.0331)
```
