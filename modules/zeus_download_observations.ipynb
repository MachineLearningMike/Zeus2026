```python
import xarray as xr

# Open NOAA OISST daily NetCDF via OpenDAP
oisst_url = "https://psl.noaa.gov/thredds/dodsC/Datasets/noaa.oisst.v2.highres/sst.day.mean.current.nc"
ds_sst = xr.open_dataset(oisst_url)
sst_today = ds_sst['sst'].sel(time='2025-12-01')
print(sst_today)
```

```python
import pandas as pd
import requests
from io import BytesIO
import gzip

# Example: download a station file (gzipped)
url = 'https://www1.ncdc.noaa.gov/pub/data/noaa/2025/010010-99999-2025.gz'
r = requests.get(url)
with gzip.open(BytesIO(r.content)) as f:
    df = pd.read_fwf(f, widths=[4,2,2,2,2,2,5], header=None)  # example widths
print(df.head())
```
